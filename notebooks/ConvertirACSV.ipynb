```python
from pathlib import Path
import pandas as pd

# Carpeta donde está el notebook y los archivos Excel
carpeta_proyecto = Path.cwd()

# Carpeta donde se guardarán los CSV convertidos
carpeta_csv = carpeta_proyecto / "csv_convertidos"
carpeta_csv.mkdir(exist_ok=True)

# Buscar archivos .xlsx y .xls
archivos_excel = (
    list(carpeta_proyecto.glob("*.xlsx")) +
    list(carpeta_proyecto.glob("*.xls"))
)

for archivo_excel in archivos_excel:
    # Ignorar archivos temporales de Excel
    if archivo_excel.name.startswith("~$"):
        continue

    libro = pd.ExcelFile(archivo_excel)

    for hoja in libro.sheet_names:
        # Leer como texto para preservar códigos e identificadores
        df = pd.read_excel(archivo_excel, sheet_name=hoja, dtype=str)

        # Si hay más de una hoja, se agrega el nombre de la hoja al CSV
        if len(libro.sheet_names) == 1:
            nombre_csv = f"{archivo_excel.stem}.csv"
        else:
            nombre_hoja = str(hoja).replace(" ", "_")
            nombre_csv = f"{archivo_excel.stem}_{nombre_hoja}.csv"

        ruta_salida = carpeta_csv / nombre_csv

        df.to_csv(ruta_salida, index=False, encoding="utf-8-sig")
        print(f"Convertido: {archivo_excel.name} → {nombre_csv}")

print("\nCSV guardados en:")
print(carpeta_csv)
```
